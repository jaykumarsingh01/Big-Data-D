```python
from ast import alias
from pyspark.sql.functions import count,avg,sum,max
df.groupBy("Country").agg(
    count("*").alias("total_people"),
    avg("age").alias("average_age"),
    sum("Experience").alias("total_experience"),
    max("age").alias("maximum_age"),
    avg("Experience").alias("average_experience")
).show()
```
