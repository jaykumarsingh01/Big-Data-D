```python
# Find the maiximum popularity and average danceability for each artist
```

```python
from ast import alias
from pyspark.sql.functions import count,avg,sum,max,min
df_pyspark.groupBy("artist").agg(

    avg("danceability").alias("average_age"),
    min("energy").alias("jay"),
    max("popularity").alias("maximum_age")

).show()
```

```python
# Find the number of tracks and average popularity for each artist sorted by track count descending
```

```python
from pyspark.sql.functions import count, avg, desc

df_pyspark.groupBy("artist").agg(
    count("*").alias("track_count"),
    avg("popularity").alias("average_popularity")
).orderBy(
    desc("track_count")
).show()
```
